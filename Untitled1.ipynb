```python
cursor.execute("CREATE DATABASE IF NOT EXISTS PChilemya")

cursor.execute("""
   CREATE TABLE IF NOT EXISTS PChilemya.authors (
       author_id INT AUTO_INCREMENT PRIMARY KEY,
       name VARCHAR(100),
       country VARCHAR(50)
   )
""")

cursor.execute("""
   CREATE TABLE IF NOT EXISTS PChilemya.books (
       book_id INT AUTO_INCREMENT PRIMARY KEY,
       title VARCHAR(150),
       year_published INT,
       genre VARCHAR(50),
       author_id INT,
       FOREIGN KEY (author_id) REFERENCES PChilemya.authors(author_id)
   )
""")

connection.commit()
print("Tables created")
```

### 1) Select and display all books

```python
cursor.execute("SELECT * FROM PChilemya.books")
print("All books:")
for row in cursor.fetchall():
    print(row)
```

### 2) Select and display only books published after 1980

```python
cursor.execute("SELECT * FROM PChilemya.books WHERE year_published > 1980")
print("\nBooks published after 1980:")
for row in cursor.fetchall():
    print(row)
```

### 3) Select and display all books by a specific author (using a placeholder)

### 4) Count how many books are in each genre

```python
cursor.execute("SELECT genre, COUNT(*) FROM PChilemya.books GROUP BY genre")
print("\nBook count by genre:")
for row in cursor.fetchall():
    print(row)
```
