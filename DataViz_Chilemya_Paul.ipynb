```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

# Ensure the dataset is loaded
cars = pd.read_csv("https://raw.githubusercontent.com/mwaskom/seaborn-data/master/mpg.csv")

plt.figure(figsize=(8, 6))
sns.boxplot(data=cars, x='cylinders', y='mpg')
plt.title('Fuel Efficiency (MPG) by Number of Cylinders')
plt.xlabel('Cylinders')
plt.ylabel('Fuel Efficiency (MPG)')
plt.show()
```

# **TASK 5:LINE CHART**




```python
mpg_by_year = cars.groupby('model_year')['mpg'].mean()
plt.figure(figsize=(10, 6))
plt.plot(mpg_by_year.index, mpg_by_year.values, marker="o")
plt.title('Average Fuel Efficiency (MPG) by Model Year')
plt.xlabel('Model Year')
plt.ylabel('Average Fuel Efficiency (MPG)')
plt.show
sns.lineplot(data = cars, x = "model_year", y = "mpg", hue = "origin", errorbar = None)

```

# **TASK 6 HEATMAP**

```python
corr = cars.corr(numeric_only=True)

# Set up a side-by-side plot comparison
fig, axes = plt.subplots(1, 2, figsize=(16, 6))

# Version 1: Original 'coolwarm'
sns.heatmap(corr, annot=True, fmt=".2f", cmap='coolwarm', vmin=-1, vmax=1, ax=axes[0])
axes[0].set_title('Correlation Heatmap (coolwarm)')

# Version 2: Alternative 'viridis'
sns.heatmap(corr, annot=True, fmt=".2f", cmap='viridis', vmin=-1, vmax=1, ax=axes[1])
axes[1].set_title('Correlation Heatmap (viridis)')

plt.tight_layout()
plt.show()
```

### **Answers to Question 1: Problems with the above chart**

1. **No Logical Order / Meaningless X-axis:** The x-axis defaults to the dataframe's index (the order in which cars were entered in the dataset). Plotting this as a line chart makes it look like noisy, chaotic time-series data when there is no chronological or sequential relationship between adjacent data rows.
2. **Misleading Visualization Choice (Line Chart):** Line charts imply a continuous trend or sequence over time. For independent observations (individual car models), a histogram, density plot, or sorted bar/scatter representation is much more appropriate.
3. **Lack of Crucial Context and Labels:** The chart lacks a title, x-axis label, and y-axis label, making it impossible for a viewer to interpret what is being measured.

### **Answer to Question 2: Corrected Meaningful Chart**
We will plot the distribution of MPG using a Kernel Density Estimate (KDE) grouped by its continent of origin to show something truly meaningful about the dataset's MPG trends.

```python
plt.figure(figsize=(10, 6))
sns.kdeplot(data=cars, x="mpg", hue="origin", fill=True, common_norm=False, palette="muted", alpha=0.5)
plt.title("Distribution of Fuel Efficiency (MPG) by Region of Origin")
plt.xlabel("Fuel Efficiency (MPG)")
plt.ylabel("Density")
plt.grid(axis='x', linestyle='--', alpha=0.5)
plt.show()
```
